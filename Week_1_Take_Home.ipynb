```python
"""5. Write a program that calculates the 
        factorial of a number using a loop.
"""

def factorialfunct(num):
    fact=1
    for i in range(1,num+1):
        
        fact *=i
        
    return fact
    

number =int(input('Enter a postive number: '))
if number < 0:
    (print(f'{number} must be a non-negative number')) 
else: 
    value =factorialfunct(number)
    print(f"{number} : {value}")

```

```python
"""6. Write a program that finds the largest number
         in a given list.
"""

number = [130,10,250,5,9,160]
min = number[0]
for i in number:
    
    if min < i:
        min =i
        
    
print(min)
```

```python
""" 7. Create a Python program that counts how many vowels
 are in a given string. """

words =input('Enter strings: ')
vowel='aeiou'
count=0
for i in words:
    if i.lower() in vowel:
        count =count+1
print('Count :',+ count)


```

```python
"""8. Write a function that takes two numbers and 
returns their sum, difference, and product."""

def funct(a,b):
    sum = a + b
    diff =a-b
    prod = a * b
    total = [sum, diff, prod]
    return total

num =funct(2,4)
print("Sum :", num[0] )
print("Difference :", num[1])
print("Product :", num[2]) 
```

```python
"""9. Create a program that reverses a string 
without using slicing (`[::-1]`)."""

string ='Animal'
reversestring = " "
for i in range (len(string)):
    reversestring =string[i]+ reversestring
print(reversestring)
```

```python
"""10. Write a program that generates the first 10 numbers of 
the Fibonacci sequence.
"""

f1= 0
f2 =1

for i in range(10):

    num =f1 + f2
    f1 =f2
    f2 =num 
print(f"Fibonnaci sequence for first 10 numbers : {num}")
    
```

```python
"""1. **Student Grades System**  
   Write a program to accept marks for 5 subjects, calculate the average, and display the grade based on:  
   - A: 80–100  
   - B: 60–79  
   - C: 40–59  
   - F: < 40  """

A= int(input("Enter scores of 1st :"))
B= int(input("Enter scores of 2nd :"))
C= int(input("Enter scores of 3rd :"))
D= int(input("Enter scores of 4th :"))
E= int(input("Enter marks of 5th :"))

Average_marks = (A+B+C+D+E)/5
print(f"The 5 subjects marks are: {A}, {B}, {C}, {D}, {E}")
if Average_marks >=80: 
    print(" A Grade")
elif Average_marks >=60:
    print(" B Grade ")
elif Average_marks >=40:
    print(" C Grade ")
else:print(" F Grade ")

```

```python
"""2. **Simple Calculator**  
   Build a calculator program that can perform addition, subtraction, 
   multiplication, and division based on user input.
"""

num_A = int(input("Enter first number :"))
num_B = int(input("Enter the second number: "))
operator = int(input("Select the appropriate number for the operation; 1. Add 2. Subtract 3. Mulitiplication 4. Division"))


if operator ==1:
    total =num_A + num_B
    print(total)
elif operator ==2:
    total =num_A - num_B
    print(total)
elif  operator==3:
    total =num_A * num_B
    print(total)
else:
    if num_B ==0:
        print("cannot divide by zero")
    else:
         total =num_A/num_B
         print(total)
```

```python
"""3. **Shopping Cart**  
   Create a program where a user can add items to a shopping cart (dictionary with item names and prices),
     then display the total cost."""
shopping_cart ={}
while True:
    item =input('Enter an item to a cart: {type q to quit}')
    if item.lower() =='q':
        break
    price =float(input('Enter the price of the item #: '))
    shopping_cart[item] =price
    
total =sum(shopping_cart.values())
print(shopping_cart)
print('Total cost :', total)
```

```python
"""4. **Guessing Game**  
   Write a program where the computer randomly selects a 
   number between 1–20, and the user must guess it. 
   Give hints if the guess is too high or too low."""
import random
generate_number =random.randint(1,20)
print(generate_number)
while True:
    Guess= int(input('Guess any number from 1-20'))
    if generate_number == Guess:
       print(f"Congratulations !! {Guess} is correct")
       break
    elif generate_number < Guess:
      print(f'{Guess} is high')
    else: print(f"{Guess} is low")


```

12. Explain the difference between **for** and **while** loops. Give an example of each. 
    
    **for loops**: are count-controlled. They are use when one already know how many times they want to repeat the code. They automatically step through a specific sequence or range. 
    **while loops**: are condition-controlled. They are use when one do not know how many times the code needs to run. They keep repeating as long as a specific condition remains true.

```python
# For loop
print("===== for loop ====")
for i in range(10):
    print(i)

# while loop
print("===== while loop ====")
i=0
while i <  10:
    print(i)
    i+=1
```
